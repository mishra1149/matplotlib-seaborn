We use distribution plots to visualize the distribution of quantitative data.

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

distplot/histplot
jointplot
pairplot
rugplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
df
```

```python
df['size'].unique() #categorical means qualtative
```

```python
sns.histplot(data=df,x='total_bill')
```

```python
sns.histplot(df['total_bill'],bins=50)
```

kernal density distribution

```python
sns.histplot(df['total_bill'],kde=True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],kde=True)
```

```python
plt.subplot(2,2,1) #second row,second column,ist position
sns.histplot(df['total_bill'])
plt.subplot(2,2,1)#second row,second column,ist position
sns.histplot(df['tip'])  
```

JOINT PLOT

```python
sns.jointplot(data=df,x='total_bill',y='tip',color='red')
```

```python
#multiple kinds we have to visualize
sns.jointplot(data=df,x='total_bill',y='tip',color='red',kind='kde')
sns.jointplot(data=df,x='total_bill',y='tip',color='red',kind='reg')
sns.jointplot(data=df,x='total_bill',y='tip',color='red',kind='hist')
sns.jointplot(data=df,x='total_bill',y='tip',color='red',kind='hex')
```

PAIRPLOT

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='sex')
#hue is all about to category devision ex:
# blue one is male and orange is female 
# total bill upar side blue so male pay own in high bills similarly tip blue is up sp in case of tip male is up
```

```python
sns.pairplot(df,hue='time')
```

```python
sns.pairplot(df,hue='size',palette='rainbow')
```

```python
sns.rugplot(df)
```
